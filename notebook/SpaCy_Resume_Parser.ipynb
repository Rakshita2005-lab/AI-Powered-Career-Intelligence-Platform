```python
def extract_text_from_pdf(file_path):

    text = ""

    with pdfplumber.open(file_path) as pdf:

        for page in pdf.pages:

            page_text = page.extract_text()

            if page_text:

                text += page_text + "\n"

    return text
```

```python
def extract_text_from_docx(file_path):

    doc = Document(file_path)

    text = ""

    for para in doc.paragraphs:

        text += para.text + "\n"

    return text
```

```python
def clean_resume(text):

    text = text.lower()

    text = re.sub(r"http\S+", "", text)

    text = re.sub(r"[^a-zA-Z ]", " ", text)

    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
skill_matcher = PhraseMatcher(nlp.vocab)

patterns = [nlp.make_doc(skill) for skill in skills]

skill_matcher.add("SKILLS", patterns)
```

```python
def extract_skills(text):

    doc = nlp(text)

    matches = skill_matcher(doc)

    extracted = []

    for match_id, start, end in matches:

        extracted.append(doc[start:end].text)

    return list(set(extracted))
```

```python
education_matcher = PhraseMatcher(nlp.vocab)

patterns = [nlp.make_doc(x) for x in education_keywords]

education_matcher.add("EDUCATION", patterns)
```

```python
def extract_education(text):

    doc = nlp(text)

    matches = education_matcher(doc)

    degrees = []

    for match_id,start,end in matches:

        degrees.append(doc[start:end].text)

    return list(set(degrees))
```

```python
def extract_experience(text):

    pattern = r'(\d+)\+?\s*(?:years|year)'

    result = re.findall(pattern,text.lower())

    if result:

        return max(result)

    return "0"
```

```python
role_matcher = PhraseMatcher(nlp.vocab)

patterns = [nlp.make_doc(x) for x in roles]

role_matcher.add("ROLE",patterns)
```

```python
def extract_role(text):

    doc = nlp(text)

    matches = role_matcher(doc)

    found=[]

    for match_id,start,end in matches:

        found.append(doc[start:end].text)

    return list(set(found))
```

```python
def parse_resume(text):

    text = clean_resume(text)

    profile = {

        "Skills": extract_skills(text),

        "Education": extract_education(text),

        "Experience": extract_experience(text),

        "Role": extract_role(text)

    }

    return profile
```
